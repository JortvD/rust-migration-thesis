```python
from datetime import datetime
import json
from src.helpers import get_project_folders, normalize_project
from src.metric import SumMetric
```

```python
INPUT_FOLDER = "/home/jortvd/thesis-data/github-results-25-3-2026"
OUTPUT_FOLDER = "../results"
```

```python
# According to CWE-1399
MEMORY_CWE = [
  "CWE-119", "CWE-120", "CWE-121", "CWE-122", "CWE-123", 
  "CWE-124", "CWE-125", "CWE-126", "CWE-127", "CWE-129", 
  "CWE-131", "CWE-193", "CWE-401", "CWE-415", "CWE-416", 
  "CWE-463", "CWE-476", "CWE-562", "CWE-587", "CWE-588", 
  "CWE-590", "CWE-762", "CWE-763", "CWE-787", "CWE-788", 
  "CWE-805", "CWE-822", "CWE-824", "CWE-825"
]

def has_memory_safety_cwe(cwes):
    for cwe in cwes:
        if cwe in MEMORY_CWE:
            return True
    return False
```

```python
metric = SumMetric("vulnerability_count")

for project_folder in get_project_folders(INPUT_FOLDER):
    print(f"Processing {project_folder.name}...")
    for release_file in sorted(project_folder.glob("advisory_*.json")):
        print(f"  Processing {release_file.name}...")
        with release_file.open() as f:
            release = json.load(f)

        if not has_memory_safety_cwe(release["cwe_ids"]):
            continue

        metric.add(
            normalize_project(project_folder.name),
            datetime.fromisoformat(release["published_at"]),
            1
        )

metric.save(OUTPUT_FOLDER)
```
